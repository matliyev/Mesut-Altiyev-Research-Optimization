## 1) Hand Solution (Derivation)

**B**

### Template for Lagrange Multipliers
1. Objective: Minimize surface area $(S(r, h) = 2\pi r h + 2\pi r^2)$.  
2. Constraint: Subject to volume $(V = \pi r^2 h = 1000 \text{cm}^3)$, with $(r>0, h>0)$.  
3. Lagrangian: $\mathcal{L}(r, h, \lambda) = 2\pi r h + 2\pi r^2 + \lambda(\pi r^2 h - 1000)$.  
4. Set partial derivatives to zero:
\begin{align}
\mathcal{L}_r &= 2\pi h + 4\pi r + \lambda (2\pi r h) = 0 \\
\mathcal{L}_h &= 2\pi r + \lambda (\pi r^2) = 0 \\
\mathcal{L}_\lambda &= \pi r^2 h - 1000 = 0
\end{align}
5. Solve for variables and classify (min/max).
\begin{align}
\lambda = \frac{2}{h} + \frac{1}{r} \\[6pt]
\lambda = \frac{2}{r} \\[6pt]
\frac{2}{h} + \frac{1}{r} = \frac{2}{r} \\[6pt]
2r = h \\[6pt]
2\pi r^3 = 1000 \text{cm}^3 \\[6pt]
r = \sqrt[3]{\frac{500}{\pi}} \\[6pt]
h = 2\sqrt[3]{\frac{500}{\pi}} \\[6pt]
\end{align}  
7. State the final answer (optimal variables and objective value).
$$
r \approx 5.419\ \text{cm} \quad \text{and} \quad h \approx 10.838\ \text{cm},
$$ 
the minimum surface area is 
$$
S_{\min} \approx 553.562\ \text{cm}^2.
$$



## 2) SymPy Solution (Symbolic)

Implement the Lagrange multiplier system in **SymPy** to confirm your hand solution.


```python

# --- SymPy template ---
import sympy as sp

r, h = sp.symbols('r h', positive = True, real = True)
lam = sp.symbols('lam', real = True)

S = 2*sp.pi*r*h + 2*sp.pi*r**2

constraint = sp.pi*r**2*h - 1000

L = S + lam*constraint

eqs = [sp.Eq(sp.diff(L, r), 0), sp.Eq(sp.diff(L, h), 0), sp.Eq(sp.diff(L, lam), 0)]

solutions = sp.solve(eqs, [r, h, lam], dict = True)
sol = solutions[0]

r_star = sp.simplify(sol[r])
h_star = sp.simplify(sol[h])
min_area = sp.simplify(S.subs({r: r_star, h: h_star}))

print("SymPy results:")
print("r* =", r_star)
print("h* =", h_star)
print("Min area =", min_area, "≈", float(min_area))

```


## 3) SciPy Solution (Numerical)

Confirm the optimum numerically with `scipy.optimize.minimize`.


```python

# --- SciPy template ---
import numpy as np
from scipy.optimize import minimize


def S_obj(v):
    r, h = v
    return 2*np.pi*r*h + 2*np.pi*r**2
def con_eq(v):
    return np.pi*v[0]**2*v[1] - 1000.0
cons = {'type': 'eq', 'fun': con_eq}
bnds = ((1e-9, None), (1e-9, None))
x0 = np.array([5.0, 12.0])
res = minimize(S_obj, x0, method='SLSQP', bounds=bnds, constraints=[cons])
res
```
