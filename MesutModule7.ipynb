My goal for this coding task is to maximize the cutoff frequency (bandwidth) of a circuit by optimizing values for R & C. I will optimize these values under certain power and cost limits to reflect real-world effects.

```python
import numpy as np
from scipy.optimize import minimize

v0 = 5         
maxPower = 0.02    
budget = 0.005  
a = 2e-6      
b = 1                
Rmin, Rmax = 1e3, 100e3
Cmin, Cmax = 0.01e-6, 10e-6

def objective(x):
    R, C = x
    return R*C

# Equality constraints force optimizer to use both limits optimally
def powerConstraint(x):
    R, C = x
    return R - v0**2 / maxPower   # must be zero

def budgetConstraint(x):
    R, C = x
    return budget - (a*R + b*C)   # must be zero

constraints = [
    {'type': 'eq', 'fun': powerConstraint},
    {'type': 'eq', 'fun': budgetConstraint}
]

bounds = [(Rmin, Rmax), (Cmin, Cmax)]

xGuess = [2000, 4e-6]

result = minimize(objective, xGuess, method='SLSQP', bounds=bounds, constraints=constraints)

print(result)

R_opt, C_opt = result.x
print("Optimal R:", R_opt)
print("Optimal C:", C_opt)
print("Minimum RC:", R_opt*C_opt)
print("Maximum cutoff frequency:", 1/(2*np.pi*R_opt*C_opt))
```

# Analysis
Did your optimal solution land on a bound (e.g., minimum R or minimum C)? If so, why does that make sense?

**My optimal solution landed on the bound of minimum R and C when my constraints were more open. However, reducing the values for my constraints produced different values for R and C. This makes sense because minimizing RC is vital to maximize the cutoff frequency, as they are inversely proportional.**

**At the optimum, R is close to V0 squared divided by max power and C is just below the line defined by (-aR + Budget)/b. Both values are near the minimum allowed by their respective constraints. This occurs because the constraints act as limits, and minimizing RC favors taking the smallest feasible R and C within these limits.**
