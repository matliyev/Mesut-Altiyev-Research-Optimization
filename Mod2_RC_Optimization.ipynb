## 1) Hand Solution (Derivation)

**Do this section first.** Type your steps clearly below using LaTeX where helpful.

1. Write the constraint and reduce to a single variable:\
$RC = T = 10 \Rightarrow R = \dfrac{10}{C}$.
2. Write cost as a single-variable function \(f(C)\):\
$f(C) = \dfrac{10}{C} + 50C$.
3. Differentiate and set to zero to find critical points:\
$f'(C) = -\dfrac{10}{C^2} + 50 = 0 \Rightarrow C^2 = \dfrac{1}{5}$.
4. Solve for $C^*$ and back-solve $R^*$. Compute minimal cost.\
$C^* = \sqrt{\dfrac{1}{5}} \approx 0.447$. $R^* = \dfrac{10}{\sqrt{\dfrac{1}{5}}} \approx 22.360$.\
$\text{Cost}^* = R^* + 50C^* \approx 22.360 + 50*0.447 \approx \$44.71$.
5. Conclude with a one-line summary of $R^*, C^*, \text{Cost}^*$.
$$R^* \approx 22.360$$\
$$C^* \approx 0.447$$\
$$\text{Cost}^* \approx \$44.71$$



## 2) SymPy Solution (Symbolic – Lagrange Multipliers)

Use SymPy to verify your hand result symbolically. Print the optimal \(R^*, C^*\) and the minimal cost.


```python

# --- SymPy solution template ---
import sympy as sp

# Symbols (positive=True helps avoid extraneous sign cases)
R, C = sp.symbols('R C', positive=True)
lam = sp.symbols('lam')
# Objective and Lagrangian
cost = R + 50*C
L = cost + lam*(R*C - 10)

eqs = [
    sp.Eq(sp.diff(L, R), 0),
    sp.Eq(sp.diff(L, C), 0),
    sp.Eq(R*C, 10)
]

sol = sp.solve(eqs, [R, C, lam], dict=True)
sol
# Extract and pretty-print
opt = sol[0]
R_star = sp.simplify(opt[R])
C_star = sp.simplify(opt[C])
min_cost = sp.simplify(cost.subs({R: R_star, C: C_star}))

print("SymPy results:")
print("R* =", R_star)
print("C* =", C_star)
print("Min Cost =", min_cost, "≈", float(min_cost))
```


## 3) SciPy Solution (Numerical – Constrained Minimize)

Use `scipy.optimize.minimize` with SLSQP, an equality constraint \(RC-10=0\), and positive bounds for \(R, C\).


```python

# --- SciPy solution template ---
import numpy as np
from scipy.optimize import minimize

def cost_fn(x):
    R, C = x
    return R + 50.0*C

# Equality constraint: R*C - 10 = 0
cons = {'type': 'eq', 'fun': lambda x: x[0]*x[1] - 10.0}

# Positive bounds for R, C
bnds = ((1e-9, None), (1e-9, None))

# Initial guess
x0 = np.array([10.0, 1.0])

res = minimize(cost_fn, x0, method='SLSQP', bounds=bnds, constraints=[cons])
res
```

```python

R_num, C_num = res.x
print(f"SciPy results:\nR* = {R_num:.6f}, C* = {C_num:.6f}")
print(f"Min Cost* = {cost_fn(res.x):.6f}")
print(f"Constraint check R*C = {R_num*C_num:.6f}")
```
